```python
# ============================================================
# ONLINE RETAIL CUSTOMER SEGMENTATION
# K-MEANS AND HIERARCHICAL CLUSTERING
# ============================================================

# ============================================================
# 1. IMPORT LIBRARIES
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans, AgglomerativeClustering
from sklearn.metrics import silhouette_score, davies_bouldin_score

from scipy.cluster.hierarchy import dendrogram, linkage

import warnings
warnings.filterwarnings("ignore")

sns.set_style("whitegrid")

print("Libraries imported successfully")


# ============================================================
# 2. LOAD DATASET
# ============================================================

# If using Kaggle:
# Add the Online Retail dataset to your notebook and change
# the path below according to the uploaded dataset.

# Excel version
df = pd.read_csv("OnlineRetail.csv", encoding='latin1')

# If your file is CSV instead, use:
# df = pd.read_csv("/kaggle/input/your-dataset/OnlineRetail.csv")

print("Dataset loaded successfully")
print("Shape:", df.shape)

df.head()


# ============================================================
# 3. BASIC INFORMATION ABOUT DATA
# ============================================================

print("\nDataset Information:")
print(df.info())

print("\nStatistical Summary:")
display(df.describe(include="all"))

print("\nColumn Names:")
print(df.columns.tolist())


# ============================================================
# 4. CHECK MISSING VALUES
# ============================================================

print("\nMissing Values:")
print(df.isnull().sum())

print("\nMissing Value Percentage:")
missing_percentage = (df.isnull().sum() / len(df)) * 100
print(missing_percentage.sort_values(ascending=False))


# ============================================================
# 5. CHECK DUPLICATES
# ============================================================

print("\nNumber of duplicate rows:")
print(df.duplicated().sum())


# ============================================================
# 6. REMOVE DUPLICATES
# ============================================================

df = df.drop_duplicates()

print("Shape after removing duplicates:", df.shape)


# ============================================================
# 7. HANDLE MISSING CUSTOMER IDs
# ============================================================

# CustomerID is necessary for customer-level clustering.
# Therefore, rows without CustomerID are removed.

df = df.dropna(subset=["CustomerID"])

print("Shape after removing missing CustomerID:", df.shape)


# ============================================================
# 8. CONVERT DATA TYPES
# ============================================================

df["CustomerID"] = df["CustomerID"].astype(int)

df["InvoiceDate"] = pd.to_datetime(df["InvoiceDate"], format='%d-%m-%Y %H:%M')

print(df.dtypes)


# ============================================================
# 9. CHECK NEGATIVE QUANTITIES
# ============================================================

print("Negative Quantity rows:", (df["Quantity"] < 0).sum())

# Negative quantity generally represents returns/cancellations.
# For customer purchase segmentation, keep only positive purchases.

df = df[df["Quantity"] > 0]

print("Shape after removing negative quantities:", df.shape)


# ============================================================
# 10. CHECK UNIT PRICE
# ============================================================

print("Zero UnitPrice rows:", (df["UnitPrice"] <= 0).sum())

df = df[df["UnitPrice"] > 0]

print("Shape after cleaning UnitPrice:", df.shape)


# ============================================================
# 11. CREATE REVENUE FEATURE
# ============================================================

df["Revenue"] = df["Quantity"] * df["UnitPrice"]

print(df[["Quantity", "UnitPrice", "Revenue"]].head())


# ============================================================
# 12. EXPLORATORY DATA ANALYSIS
# ============================================================

# Top countries by number of transactions

country_orders = df["Country"].value_counts().head(10)

plt.figure(figsize=(12, 6))
country_orders.plot(kind="bar", color="steelblue")
plt.title("Top 10 Countries by Number of Transactions")
plt.xlabel("Country")
plt.ylabel("Number of Transactions")
plt.xticks(rotation=45)
plt.show()


# ============================================================
# 13. TOP COUNTRIES BY REVENUE
# ============================================================

country_revenue = (
    df.groupby("Country")["Revenue"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

plt.figure(figsize=(12, 6))
country_revenue.plot(kind="bar", color="darkorange")
plt.title("Top 10 Countries by Revenue")
plt.xlabel("Country")
plt.ylabel("Revenue")
plt.xticks(rotation=45)
plt.show()


# ============================================================
# 14. MONTHLY REVENUE
# ============================================================

df["YearMonth"] = df["InvoiceDate"].dt.to_period("M")

monthly_revenue = df.groupby("YearMonth")["Revenue"].sum()

plt.figure(figsize=(14, 6))
monthly_revenue.plot(color="green", marker="o")
plt.title("Monthly Revenue")
plt.xlabel("Month")
plt.ylabel("Revenue")
plt.xticks(rotation=45)
plt.show()


# ============================================================
# 15. REVENUE DISTRIBUTION
# ============================================================

plt.figure(figsize=(10, 5))
sns.boxplot(x=df["Revenue"])
plt.title("Revenue Distribution")
plt.show()


# ============================================================
# 16. RFM ANALYSIS
# ============================================================

# R = Recency
# F = Frequency
# M = Monetary

# Reference date is one day after the latest transaction.

reference_date = df["InvoiceDate"].max() + pd.Timedelta(days=1)

print("Reference Date:", reference_date)


# ============================================================
# 17. CALCULATE RECENCY
# ============================================================

recency_df = (
    df.groupby("CustomerID")["InvoiceDate"]
    .max()
    .reset_index()
)

recency_df["Recency"] = (
    reference_date - recency_df["InvoiceDate"]
).dt.days

recency_df = recency_df[["CustomerID", "Recency"]]

recency_df.head()


# ============================================================
# 18. CALCULATE FREQUENCY
# ============================================================

frequency_df = (
    df.groupby("CustomerID")["InvoiceNo"]
    .nunique()
    .reset_index()
)

frequency_df.columns = ["CustomerID", "Frequency"]

frequency_df.head()


# ============================================================
# 19. CALCULATE MONETARY VALUE
# ============================================================

monetary_df = (
    df.groupby("CustomerID")["Revenue"]
    .sum()
    .reset_index()
)

monetary_df.columns = ["CustomerID", "Monetary"]

monetary_df.head()


# ============================================================
# 20. COMBINE RFM FEATURES
# ============================================================

rfm = recency_df.merge(
    frequency_df,
    on="CustomerID"
)

rfm = rfm.merge(
    monetary_df,
    on="CustomerID"
)

print("RFM Dataset Shape:", rfm.shape)

display(rfm.head())


# ============================================================
# 21. RFM STATISTICS
# ============================================================

print("\nRFM Statistics:")
display(rfm[["Recency", "Frequency", "Monetary"]].describe())


# ============================================================
# 22. VISUALIZE RFM FEATURES
# ============================================================

fig, axes = plt.subplots(1, 3, figsize=(18, 5))

sns.histplot(rfm["Recency"], bins=30, kde=True, ax=axes[0])
axes[0].set_title("Recency Distribution")

sns.histplot(rfm["Frequency"], bins=30, kde=True, ax=axes[1])
axes[1].set_title("Frequency Distribution")

sns.histplot(rfm["Monetary"], bins=30, kde=True, ax=axes[2])
axes[2].set_title("Monetary Distribution")

plt.tight_layout()
plt.show()


# ============================================================
# 23. CHECK OUTLIERS
# ============================================================

fig, axes = plt.subplots(1, 3, figsize=(18, 5))

sns.boxplot(y=rfm["Recency"], ax=axes[0])
axes[0].set_title("Recency Outliers")

sns.boxplot(y=rfm["Frequency"], ax=axes[1])
axes[1].set_title("Frequency Outliers")

sns.boxplot(y=rfm["Monetary"], ax=axes[2])
axes[2].set_title("Monetary Outliers")

plt.tight_layout()
plt.show()


# ============================================================
# 24. REMOVE OUTLIERS USING IQR
# ============================================================

def remove_outliers_iqr(data, column):
    Q1 = data[column].quantile(0.25)
    Q3 = data[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    return data[
        (data[column] >= lower_limit) &
        (data[column] <= upper_limit)
    ]


rfm_clean = rfm.copy()

rfm_clean = remove_outliers_iqr(rfm_clean, "Recency")
rfm_clean = remove_outliers_iqr(rfm_clean, "Frequency")
rfm_clean = remove_outliers_iqr(rfm_clean, "Monetary")

print("Original RFM shape:", rfm.shape)
print("After outlier removal:", rfm_clean.shape)


# ============================================================
# 25. LOG TRANSFORMATION
# ============================================================

# Log transformation reduces skewness.

rfm_log = rfm_clean[["Recency", "Frequency", "Monetary"]].copy()

rfm_log = np.log1p(rfm_log)

print("Log-transformed data:")
display(rfm_log.head())


# ============================================================
# 26. STANDARDIZATION
# ============================================================

scaler = StandardScaler()

rfm_scaled = scaler.fit_transform(rfm_log)

rfm_scaled = pd.DataFrame(
    rfm_scaled,
    columns=["Recency", "Frequency", "Monetary"],
    index=rfm_clean.index
)

print("Scaled RFM dataेंट्स:")
display(rfm_scaled.head())


# ============================================================
# 27. CORRELATION MATRIX
# ============================================================

plt.figure(figsize=(8, 6))

sns.heatmap(
    rfm_scaled.corr(),
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Matrix of RFM Features")
plt.show()


# ============================================================
# 28. K-MEANS - ELBOW METHOD
# ============================================================

inertia = []

K = range(2, 11)

for k in K:
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    kmeans.fit(rfm_scaled)

    inertia.append(kmeans.inertia_)


plt.figure(figsize=(10, 6))

plt.plot(
    K,
    inertia,
    marker="o",
    color="blue"
)

plt.title("Elbow Method for Optimal Number of Clusters")
plt.xlabel("Number of Clusters (K)")
plt.ylabel("Inertia")
plt.xticks(K)
plt.show()


# ============================================================
# 29. SILHOUETTE SCORE FOR DIFFERENT K
# ============================================================

silhouette_scores = []

for k in K:

    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    labels = kmeans.fit_predict(rfm_scaled)

    score = silhouette_score(
        rfm_scaled,
        labels
    )

    silhouette_scores.append(score)


plt.figure(figsize=(10, 6))

plt.plot(
    K,
    silhouette_scores,
    marker="o",
    color="green"
)

plt.title("Silhouette Score for Different K Values")
plt.xlabel("Number of Clusters")
plt.ylabel("Silhouette Score")
plt.xticks(K)
plt.show()


# ============================================================
# 30. DAVIES-BOULDIN INDEX
# ============================================================

db_scores = []

for k in K:

    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    labels = kmeans.fit_predict(rfm_scaled)

    score = davies_bouldin_score(
        rfm_scaled,
        labels
    )

    db_scores.append(score)


plt.figure(figsize=(10, 6))

plt.plot(
    K,
    db_scores,
    marker="o",
    color="red"
)

plt.title("Davies-Bouldin Index for Different K Values")
plt.xlabel("Number of Clusters")
plt.ylabel("Davies-Bouldin Index")
plt.xticks(K)
plt.show()


# ============================================================
# 31. SELECT NUMBER OF CLUSTERS
# ============================================================

# You can change this value based on the Elbow and
# Silhouette plots.

optimal_k = 3

print("Selected number of clusters:", optimal_k)


# ============================================================
# 32. BUILD FINAL K-MEANS MODEL
# ============================================================

kmeans = KMeans(
    n_clusters=optimal_k,
    random_state=42,
    n_init=10
)

rfm_clean["KMeans_Cluster"] = kmeans.fit_predict(rfm_scaled)

print("K-Means clustering completed.")


# ============================================================
# 33. K-MEANS SILHOUETTE SCORE
# ============================================================

kmeans_silhouette = silhouette_score(
    rfm_scaled,
    rfm_clean["KMeans_Cluster"]
)

print("K-Means Silhouette Score:",
      round(kmeans_silhouette, 4))


# ============================================================
# 34. K-MEANS DAVIES-BOULDIN SCORE
# ============================================================

kmeans_db = davies_bouldin_score(
    rfm_scaled,
    rfm_clean["KMeans_Cluster"]
)

print("K-Means Davies-Bouldin Score:",
      round(kmeans_db, 4))


# ============================================================
# 35. CLUSTER DISTRIBUTION
# ============================================================

cluster_counts = rfm_clean["KMeans_Cluster"].value_counts().sort_index()

print(cluster_counts)

plt.figure(figsize=(8, 5))

sns.barplot(
    x=cluster_counts.index,
    y=cluster_counts.values,
    palette="viridis"
)

plt.title("Number of Customers in Each K-Means Cluster")
plt.xlabel("Cluster")
plt.ylabel("Number of Customers")
plt.show()


# ============================================================
# 36. CLUSTER PROFILE
# ============================================================

cluster_profile = (
    rfm_clean
    .groupby("KMeans_Cluster")[["Recency", "Frequency", "Monetary"]]
    .mean()
    .round(2)
)

print("K-Means Cluster Profile:")
display(cluster_profile)


# ============================================================
# 37. MEDIAN CLUSTER PROFILE
# ============================================================

cluster_median = (
    rfm_clean
    .groupby("KMeans_Cluster")[["Recency", "Frequency", "Monetary"]]
    .median()
    .round(2)
)

print("Median RFM values by cluster:")
display(cluster_median)


# ============================================================
# 38. VISUALIZE K-MEANS CLUSTERS
# ============================================================

plt.figure(figsize=(10, 7))

sns.scatterplot(
    data=rfm_clean,
    x="Frequency",
    y="Monetary",
    hue="KMeans_Cluster",
    palette="viridis",
    s=80
)

plt.title("K-Means Customer Segmentation")
plt.xlabel("Frequency")
plt.ylabel("Monetary Value")
plt.legend(title="Cluster")
plt.show()


# ============================================================
# 39. HIERARCHICAL CLUSTERING - DENDROGRAM
# ============================================================

# Ward linkage is commonly used for hierarchical clustering.

linked = linkage(
    rfm_scaled,
    method="ward"
)

plt.figure(figsize=(15, 7))

dendrogram(
    linked,
    truncate_mode="lastp",
    p=30,
    leaf_rotation=90,
    leaf_font_size=10,
    show_contracted=True
)

plt.title("Hierarchical Clustering Dendrogram")
plt.xlabel("Customer Groups")
plt.ylabel("Distance")

plt.show()


# ============================================================
# 40. AGGLOMERATIVE HIERARCHICAL CLUSTERING
# ============================================================

hierarchical = AgglomerativeClustering(
    n_clusters=optimal_k,
    linkage="ward"
)

hierarchical_labels = hierarchical.fit_predict(
    rfm_scaled
)

rfm_clean["Hierarchical_Cluster"] = hierarchical_labels

print("Hierarchical clustering completed.")


# ============================================================
# 41. HIERARCHICAL SILHOUETTE SCORE
# ============================================================

hierarchical_silhouette = silhouette_score(
    rfm_scaled,
    hierarchical_labels
)

print(
    "Hierarchical Silhouette Score:",
    round(hierarchical_silhouette, 4)
)


# ============================================================
# 42. HIERARCHICAL DAVIES-BOULDIN SCORE
# ============================================================

hierarchical_db = davies_bouldin_score(
    rfm_scaled,
    hierarchical_labels
)

print(
    "Hierarchical Davies-Bouldin Score:",
    round(hierarchical_db, 4)
)


# ============================================================
# 43. HIERARCHICAL CLUSTER DISTRIBUTION
# ============================================================

hierarchical_counts = (
    rfm_clean["Hierarchical_Cluster"]
    .value_counts()
    .sort_index()
)

print(hierarchical_counts)

plt.figure(figsize=(8, 5))

sns.barplot(
    x=hierarchical_counts.index,
    y=hierarchical_counts.values,
    palette="Set2"
)

plt.title("Hierarchical Cluster Distribution")
plt.xlabel("Cluster")
plt.ylabel("Number of Customers")
plt.show()


# ============================================================
# 44. HIERARCHICAL CLUSTER PROFILE
# ============================================================

hierarchical_profile = (
    rfm_clean
    .groupby("Hierarchical_Cluster")
    [["Recency", "Frequency", "Monetary"]]
    .mean()
    .round(2)
)

print("Hierarchical Cluster Profile:")
display(hierarchical_profile)


# ============================================================
# 45. VISUALIZE HIERARCHICAL CLUSTERS
# ============================================================

plt.figure(figsize=(10, 7))

sns.scatterplot(
    data=rfm_clean,
    x="Frequency",
    y="Monetary",
    hue="Hierarchical_Cluster",
    palette="Set2",
    s=80
)

plt.title("Hierarchical Customer Segmentation")
plt.xlabel("Frequency")
plt.ylabel("Monetary Value")
plt.legend(title="Cluster")
plt.show()


# ============================================================
# 46. COMPARE BOTH ALGORITHMS
# ============================================================

comparison = pd.DataFrame({
    "Algorithm": [
        "K-Means",
        "Hierarchical Clustering"
    ],

    "Silhouette Score": [
        kmeans_silhouette,
        hierarchical_silhouette
    ],

    "Davies-Bouldin Score": [
        kmeans_db,
        hierarchical_db
    ]
})

comparison["Silhouette Score"] = comparison["Silhouette Score"].round(4)
comparison["Davies-Bouldin Score"] = comparison["Davies-Bouldin Score"].round(4)

display(comparison)


# ============================================================
# 47. CREATE CUSTOMER SEGMENT LABELS
# ============================================================

# Find the average RFM values for each cluster.

profile = (
    rfm_clean
    .groupby("KMeans_Cluster")
    [["Recency", "Frequency", "Monetary"]]
    .mean()
)

print(profile)


# ============================================================
# 48. AUTOMATICALLY LABEL K-MEANS CLUSTERS
# ============================================================

# High Frequency + High Monetary + Low Recency = Loyal
# Low Frequency + Low Monetary + High Recency = At Risk
# Others = Potential

cluster_labels = {}

for cluster in profile.index:

    recency = profile.loc[cluster, "Recency"]
    frequency = profile.loc[cluster, "Frequency"]
    monetary = profile.loc[cluster, "Monetary"]

    if (
        frequency >= profile["Frequency"].median()
        and monetary >= profile["Monetary"].median()
        and recency <= profile["Recency"].median()
    ):
        cluster_labels[cluster] = "Loyal Customers"

    elif (
        recency >= profile["Recency"].median()
        and frequency <= profile["Frequency"].median()
    ):
        cluster_labels[cluster] = "At Risk Customers"

    else:
        cluster_labels[cluster] = "Potential Customers"


print("Cluster Labels:")
print(cluster_labels)


# ============================================================
# 49. ADD CUSTOMER SEGMENT
# ============================================================

rfm_clean["Customer_Segment"] = (
    rfm_clean["KMeans_Cluster"]
    .map(cluster_labels)
)

display(
    rfm_clean[
        [
            "CustomerID",
            "Recency",
            "Frequency",
            "Monetary",
            "KMeans_Cluster",
            "Customer_Segment"
        ]
    ].head(20)
)


# ============================================================
# 50. CUSTOMER SEGMENT DISTRIBUTION
# ============================================================

segment_counts = rfm_clean["Customer_Segment"].value_counts()

plt.figure(figsize=(10, 6))

sns.barplot(
    x=segment_counts.index,
    y=segment_counts.values,
    palette="Set1"
)

plt.title("Customer Segment Distribution")
plt.xlabel("Customer Segment")
plt.ylabel("Number of Customers")
plt.xticks(rotation=20)

plt.show()


# ============================================================
# 51. SEGMENT PROFILE
# ============================================================

segment_profile = (
    rfm_clean
    .groupby("Customer_Segment")
    [["Recency", "Frequency", "Monetary"]]
    .mean()
    .round(2)
)

display(segment_profile)


# ============================================================
# 52. CUSTOMER SEGMENT PIE CHART
# ============================================================

plt.figure(figsize=(8, 8))

plt.pie(
    segment_counts.values,
    labels=segment_counts.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Customer Segmentation Distribution")

plt.show()


# ============================================================
# 53. 3D VISUALIZATION OF RFM CLUSTERS
# ============================================================

from mpl_toolkits.mplot3d import Axes3D

fig = plt.figure(figsize=(12, 8))

ax = fig.add_subplot(111, projection="3d")

scatter = ax.scatter(
    rfm_clean["Recency"],
    rfm_clean["Frequency"],
    rfm_clean["Monetary"],
    c=rfm_clean["KMeans_Cluster"],
    cmap="viridis",
    s=40
)

ax.set_xlabel("Recency")
ax.set_ylabel("Frequency")
ax.set_zlabel("Monetary")

ax.set_title("3D Visualization of Customer Clusters")

plt.show()


# ============================================================
# 54. SAVE RESULTS
# ============================================================

output_file = "/kaggle/working/customer_segmentation_results.csv"

rfm_clean.to_csv(
    output_file,
    index=False
)

print("Results saved to:")
print(output_file)


# ============================================================
# 55. FINAL RESULTS
# ============================================================

print("=" * 60)
print("FINAL CLUSTERING RESULTS")
print("=" * 60)

print("\nNumber of customers analyzed:",
      len(rfm_clean))

print("\nNumber of clusters:",
      optimal_k)

print("\nK-Means Silhouette Score:",
      round(kmeans_silhouette, 4))

print("\nK-Means Davies-Bouldin Score:",
      round(kmeans_db, 4))

print("\nHierarchical Silhouette Score:",
      round(hierarchical_silhouette, 4))

print("\nHierarchical Davies-Bouldin Score:",
      round(hierarchical_db, 4))

print("\nCustomer Segments:")
print(rfm_clean["Customer_Segment"].value_counts())

print("\nProject completed successfully!")
```
